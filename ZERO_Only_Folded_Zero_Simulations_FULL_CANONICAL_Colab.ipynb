## Suite A — finite pair closure and increasing boundary [E]

From the canonical finite expression $0_N=0+\sum_{n=1}^N(n+(-n))=0$, evaluate finite inverse pairs without changing the source $0_Q$ expression. The manuscript tests positive finite $N$; $N=0$ is included separately as the auxiliary empty-sum edge case and is **not** a replacement definition.

```python
def folded_zero_finite(N):
    # [E] Evaluates 0_N, NOT 0_Q.
    N = require_integer(N)
    if N < 0: raise ValueError('N must be nonnegative')
    n = np.arange(1,N+1,dtype=np.int64)
    pair_sums = n+(-n)
    return {'N':N, 'pair_count':N, 'represented_finite_integers':2*N+1,
            'numeric_0_N':int(pair_sums.sum()),
            'every_pair_closes':bool(np.all(pair_sums==0))}

POSITIVE_BOUNDARIES=[1,2,10,100,1000,10000,100000,1000000]
finite_results=pd.DataFrame([folded_zero_finite(N) for N in POSITIVE_BOUNDARIES])
assert finite_results['every_pair_closes'].all()
assert (finite_results['numeric_0_N']==0).all()
assert (finite_results['represented_finite_integers']==2*finite_results['N']+1).all()
display(finite_results)
print('N=0 auxiliary empty-sum check:', folded_zero_finite(0))
```

```python
def pair_closure_table(N=10):
    N=require_integer(N)
    if N<0: raise ValueError('N must be nonnegative')
    n=np.arange(1,N+1,dtype=np.int64)
    return pd.DataFrame({'n':n,'+n':n,'-n':-n,'n+(-n)':n+(-n)})

display(pair_closure_table(12))
assert (pair_closure_table(12)['n+(-n)']==0).all()
```

```python
def boundary_growth(max_power=6):
    max_power=require_integer(max_power,'max_power')
    if not 0<=max_power<=6: raise ValueError('choose a maximum power between 0 and 6')
    vals=[10**p for p in range(max_power+1)]
    return pd.DataFrame([folded_zero_finite(N) for N in vals])

growth=boundary_growth(6)
fig,ax=plt.subplots(1,2,figsize=(12,4))
ax[0].loglog(growth['N'],growth['represented_finite_integers'],marker='o')
ax[0].set(title='Represented finite state count [E]',xlabel='Finite boundary N',ylabel='2N+1')
ax[1].semilogx(growth['N'],growth['numeric_0_N'],marker='o')
ax[1].set(title='Finite inverse-pair closure [E]',xlabel='Finite boundary N',ylabel='0_N')
for a in ax:a.grid(alpha=.25)
plt.tight_layout();plt.show()
display(growth)
```

The next two displays reproduce the source experiment's number-line and paired-location ideas. They are **drawings of finite inverse pairs**, not an external geometric model of Zero, not an ordering of the $\parallel$ coexistence, and not an extra coordinate of the Echo Meridian.

```python
def plot_number_line(N=10):
    N=require_integer(N)
    if N<1:raise ValueError('Use N>=1 to plot symmetric finite values')
    values=np.arange(-N,N+1)
    fig,ax=plt.subplots(figsize=(12,2.5))
    ax.axhline(0,lw=.9,color='gray')
    ax.scatter(values,np.zeros_like(values),s=35)
    ax.scatter([0],[0],s=120,marker='o')
    ax.set(xlim=(-N-1,N+1),ylim=(-1,1),yticks=[],xlabel='Finite integers (analysis display)',title=f'Finite symmetric window [-{N}, {N}] [A]')
    ax.grid(axis='x',alpha=.22)
    plt.tight_layout();plt.show()

plot_number_line(10)
```

```python
def folded_coordinates(N):
    N=require_integer(N)
    if N<1:raise ValueError('N>=1 is required for finite illustrative pairs')
    n=np.arange(1,N+1)
    angles=2*np.pi*(n-1)/N
    radii=n/N
    xp=radii*np.cos(angles);yp=radii*np.sin(angles)
    return {'n':n,'xp':xp,'yp':yp,'xn':-xp,'yn':-yp}

def plot_folded_zero(N=12):
    pts=folded_coordinates(N)
    fig,ax=plt.subplots(figsize=(7,7))
    for x,y,xx,yy in zip(pts['xp'],pts['yp'],pts['xn'],pts['yn']):
        ax.plot([x,0,xx],[y,0,yy],lw=.9,alpha=.65)
    ax.scatter(pts['xp'],pts['yp'],label='+n [finite display]')
    ax.scatter(pts['xn'],pts['yn'],label='-n [finite display]')
    ax.scatter([0],[0],s=140,label='Zero reference (schematic)')
    for v,x,y in zip(pts['n'],pts['xp'],pts['yp']):ax.annotate(f'+{v}',(x,y),xytext=(3,3),textcoords='offset points',fontsize=8)
    for v,x,y in zip(pts['n'],pts['xn'],pts['yn']):ax.annotate(f'-{v}',(x,y),xytext=(3,-8),textcoords='offset points',fontsize=8)
    ax.axis('equal');ax.axis('off');ax.set_title(f'Finite paired-state schematic [A] | N={N}')
    ax.legend(loc='upper left');plt.tight_layout();plt.show()

plot_folded_zero(12)
```

## Suite D — independent finite audit and source-fidelity checks

This independent check recalculates the finite pair-sum identity without treating its scalar result as the full $0_Q$ form. The full equation and conditions stay in the source registry, and the return relation uses one formal Zero reference.

```python
def independent_finite_check(N):
    N=require_integer(N)
    if N<0:raise ValueError('N must be nonnegative')
    # [E] Integer exactness from pair-by-pair arithmetic; not an infinite-state simulation.
    pairs=[n+(-n) for n in range(1,N+1)]
    return (sum(pairs)==0 and all(pair==0 for pair in pairs),2*N+1)

for N in (0,1,3,10,200,1000):
    closed,count=independent_finite_check(N)
    assert closed and count==folded_zero_finite(N)['represented_finite_integers']
    assert count==2*N+1
print('PASS: independent finite scalar calculation matches all checked boundaries.')
```
